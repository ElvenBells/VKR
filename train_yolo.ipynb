```python
from ultralytics import YOLO

# Load a model
model = YOLO('yolov8m.pt')  # load a pretrained model (recommended for training)

# Train the model
results = model.train(data='my_dataset_yolo/data.yaml',
                      batch=4,
                      workers=0,
                      cache=False,
                      device=0,
                      epochs=50,
                      imgsz=640,
                      model='yolov8m.pt',
                      amp=True,
                      verbose=True,
                      project="models",
                      name="experiment",
                      exist_ok=True
                      )
```
